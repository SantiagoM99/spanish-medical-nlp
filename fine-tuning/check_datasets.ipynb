```python
import pandas as pd

anat = pd.read_json('results/anat_em_beto.json')
anat_robertas = pd.read_json('results/anat_em_robertas.json')
anat_robertas_2 = pd.read_json('results/anat_em_robertas_2.json')
anat_sci = pd.read_json('results/anat_em_robertas_sci.json')
anat_xlm = pd.read_json('results/anat_em_robertas_xlm.json')

anat = pd.concat([anat, anat_robertas], ignore_index=True)
anat = pd.concat([anat, anat_robertas_2], ignore_index=True)
anat = pd.concat([anat, anat_sci], ignore_index=True)
anat = pd.concat([anat, anat_xlm], ignore_index=True)

anat = anat.sort_values(by="test_f1", ascending=False).reset_index(drop=True)
anat
```
